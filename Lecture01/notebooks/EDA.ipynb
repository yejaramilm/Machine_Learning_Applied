```python
import pandas as pd
```

```python
#AGRUPACIÓN
datos = {
    'id': [1, 2, 3, 4, 5],
    'categoria': ['Electronica', 'Hogar', 'Electronica', 'Hogar', 'Ropa'],
    'ventas': [800, 150, 1200, 350, 100],
    'producto': ['Laptop', 'Lampara', 'Smartphone', 'Silla', 'Camiseta']
}

# Convertimos el diccionario en un DataFrame de Pandas
df = pd.DataFrame(datos)
df
```

```python
# Ejecutamos el cálculo del promedio por categoría
promedio_ventas = df.groupby('categoria')['ventas'].mean()
print(promedio_ventas)
```
